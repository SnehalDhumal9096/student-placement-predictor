```python
model = LogisticRegression(max_iter=1000)

model.fit(x_train_processed, y_train)

y_pred = model.predict(x_test_processed)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# Create Decision Tree model
dt_model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

# Train the model
dt_model.fit(x_train_processed, y_train)

# Make predictions
dt_pred = dt_model.predict(x_test_processed)

# Evaluate the model
print("Decision Tree Accuracy:", accuracy_score(y_test, dt_pred))

print("\nClassification Report:")
print(classification_report(y_test, dt_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=5,
    random_state=42
)

# Train the model
rf_model.fit(x_train_processed, y_train)

# Make predictions
rf_pred = rf_model.predict(x_test_processed)

# Evaluate the model
print("Random Forest Accuracy:", accuracy_score(y_test, rf_pred))

print("\nClassification Report:")
print(classification_report(y_test, rf_pred))
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

# Remove target, leakage columns, and student ID
x = df.drop(
    ["placed", "package_lpa", "company_type", "student_id"],
    axis=1
)
y = df["placed"]

# Split data
x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42
)

# Identify columns
categorical_cols = x.select_dtypes(include="object").columns
numerical_cols = x.select_dtypes(exclude="object").columns

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numerical_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ]
)

# Transform data
x_train_processed = preprocessor.fit_transform(x_train)
x_test_processed = preprocessor.transform(x_test)

# Random Forest
rf = RandomForestClassifier(random_state=42)

# Parameters to test
param_grid = {
    "n_estimators": [50, 100, 200],
    "max_depth": [3, 5, 7, None]
}

# Grid Search
grid_search = GridSearchCV(
    rf,
    param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

# Train and find best parameters
grid_search.fit(x_train_processed, y_train)

# Best model
best_rf = grid_search.best_estimator_

# Predictions
best_rf_pred = best_rf.predict(x_test_processed)

# Results
print("Best Parameters:", grid_search.best_params_)
print("Best Cross-Validation Accuracy:", grid_search.best_score_)
print("Test Accuracy:", accuracy_score(y_test, best_rf_pred))

print("\nClassification Report:")
print(classification_report(y_test, best_rf_pred))
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Create confusion matrix
cm = confusion_matrix(y_test, best_rf_pred)

print("Confusion Matrix:")
print(cm)

# Visualize
plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=["Not Placed", "Placed"],
    yticklabels=["Not Placed", "Placed"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - Final Random Forest")
plt.show()
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import joblib

# Load saved model and preprocessor
model = joblib.load("placement_model.pkl")
preprocessor = joblib.load("placement_preprocessor.pkl")

st.title("🎓 Student Placement Prediction System")
st.write("Predict whether a student is likely to be placed.")

# Student inputs
gender = st.selectbox("Gender", ["Male", "Female"])
age = st.number_input("Age", min_value=18, max_value=30, value=22)

degree = st.selectbox(
    "Degree",
    ["BCA", "BE", "BSc", "BTech"]
)

branch = st.selectbox(
    "Branch",
    ["AI", "CS", "DS", "Electrical", "IT", "Mechanical"]
)

backlogs = st.number_input(
    "Number of Backlogs",
    min_value=0,
    max_value=10,
    value=0
)

internships = st.number_input(
    "Number of Internships",
    min_value=0,
    max_value=10,
    value=1
)

certifications = st.number_input(
    "Number of Certifications",
    min_value=0,
    max_value=10,
    value=2
)

coding_skills = st.slider(
    "Coding Skills",
    min_value=1,
    max_value=10,
    value=5
)

communication_skills = st.slider(
    "Communication Skills",
    min_value=1,
    max_value=10,
    value=5
)

projects = st.number_input(
    "Number of Projects",
    min_value=0,
    max_value=10,
    value=2
)

aptitude_score = st.number_input(
    "Aptitude Score",
    min_value=0.0,
    max_value=100.0,
    value=50.0
)

cgpa = st.number_input(
    "CGPA",
    min_value=0.0,
    max_value=10.0,
    value=7.5
)

# Prediction button
if st.button("Predict Placement"):

    new_student = pd.DataFrame({
        "gender": [gender],
        "age": [age],
        "degree": [degree],
        "branch": [branch],
        "backlogs": [backlogs],
        "internships": [internships],
        "certifications": [certifications],
        "coding_skills": [coding_skills],
        "communication_skills": [communication_skills],
        "projects": [projects],
        "aptitude_score": [aptitude_score],
        "cgpa": [cgpa]
    })

    # Preprocess
    new_student_processed = preprocessor.transform(new_student)

    # Prediction
    prediction = model.predict(new_student_processed)
    probability = model.predict_proba(new_student_processed)

    st.subheader("Prediction Result")

    if prediction[0] == 1:
        st.success("🎉 Student is predicted to be PLACED")
    else:
        st.error("❌ Student is predicted to be NOT PLACED")

    st.write(
        f"Probability of Placement: "
        f"{probability[0][1] * 100:.2f}%"
    )

    st.write(
        f"Probability of Not Placement: "
        f"{probability[0][0] * 100:.2f}%"
    )
```

```python
# ============================================================
# FINAL MODEL EVALUATION — CORRECTED
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)
import pandas as pd
import json
import os

PROJECT = "/content/student_placement_project"

# ------------------------------------------------------------
# 1. CREATE ORIGINAL X
# ------------------------------------------------------------

X_full = df.drop(columns=["placed"])

# ------------------------------------------------------------
# 2. RECOVER EXACT TEST DATA
# ------------------------------------------------------------

X_test = X_full.loc[y_test.index]

print("✅ X_test recovered")
print("X_test shape:", X_test.shape)

# ------------------------------------------------------------
# 3. APPLY THE SAME PREPROCESSING USED DURING TRAINING
# ------------------------------------------------------------

X_test_processed = preprocessor.transform(X_test)

print("✅ X_test preprocessing completed")

print(
    "Processed X_test shape:",
    X_test_processed.shape
)

# ------------------------------------------------------------
# 4. PREDICT USING RANDOM FOREST
# ------------------------------------------------------------

y_pred = best_rf.predict(X_test_processed)

print("✅ Predictions generated")

# ------------------------------------------------------------
# 5. CALCULATE METRICS
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

# ------------------------------------------------------------
# 6. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("📊 FINAL RANDOM FOREST MODEL RESULTS")
print("=" * 60)

print(f"\nAccuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1 Score  : {f1 * 100:.2f}%")

# ------------------------------------------------------------
# 7. CLASSIFICATION REPORT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("📋 CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)

# ------------------------------------------------------------
# 8. SAVE METRICS
# ------------------------------------------------------------

metrics = {
    "accuracy": round(float(accuracy), 4),
    "precision": round(float(precision), 4),
    "recall": round(float(recall), 4),
    "f1_score": round(float(f1), 4)
}

metrics_path = os.path.join(
    PROJECT,
    "model_metrics.json"
)

with open(
    metrics_path,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        metrics,
        file,
        indent=4
    )

print("\n✅ Metrics saved to:")
print(metrics_path)

# ------------------------------------------------------------
# 9. SAVE PREDICTIONS
# ------------------------------------------------------------

results = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

results.to_csv(
    os.path.join(
        PROJECT,
        "model_predictions.csv"
    ),
    index=False
)

print("✅ Predictions saved.")

print("\n" + "=" * 60)
print("🎉 FINAL MODEL EVALUATION COMPLETED")
print("=" * 60)
```
